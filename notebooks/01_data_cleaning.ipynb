# 01 Data Cleaning & Preprocessing
This notebook performs data cleaning, missing value handling, datatype conversion, and consistency checks on the pest, weather, and crop yield datasets.


```python
import pandas as pd
import numpy as np
import os

# Load Datasets
pest_weather_df = pd.read_csv('../datasets/pest_weather.csv')
crop_yield_df = pd.read_csv('../datasets/crop_yield.csv')

print('Pest Weather Dataset Shape:', pest_weather_df.shape)
print('Crop Yield Dataset Shape:', crop_yield_df.shape)
```

```python
# Check missing values
print('--- Pest Weather Missing Values ---')
print(pest_weather_df.isnull().sum())

print('\n--- Crop Yield Missing Values ---')
print(crop_yield_df.isnull().sum())
```

```python
# Data Cleaning and Handling Duplicates
pest_weather_clean = pest_weather_df.drop_duplicates()
crop_yield_clean = crop_yield_df.drop_duplicates()

print('Duplicates removed in pest_weather:', len(pest_weather_df) - len(pest_weather_clean))
print('Duplicates removed in crop_yield:', len(crop_yield_df) - len(crop_yield_clean))
```

```python
# Data Type Conversions & Outlier Checks
print(pest_weather_clean.describe())
```
