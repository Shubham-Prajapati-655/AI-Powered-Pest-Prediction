# 02 Exploratory Data Analysis (EDA)
This notebook explores weather patterns, pest outbreak frequencies, crop distributions, and relationships between climate parameters and pest risks.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
pest_weather = pd.read_csv('../datasets/pest_weather.csv')
crop_yield = pd.read_csv('../datasets/crop_yield.csv')

pest_weather.head()
```

```python
# Distribution of Pest Outbreaks
plt.figure(figsize=(8, 4))
sns.countplot(data=pest_weather, x='Outbreak', palette='viridis', order=['Low', 'Medium', 'High'])
plt.title('Distribution of Pest Outbreak Severity Levels')
plt.show()
```

```python
# Weather distribution vs Outbreak
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.boxplot(data=pest_weather, x='Outbreak', y='Temperature', ax=axes[0], palette='Blues')
axes[0].set_title('Temperature vs Outbreak')
sns.boxplot(data=pest_weather, x='Outbreak', y='Humidity', ax=axes[1], palette='Greens')
axes[1].set_title('Humidity vs Outbreak')
sns.boxplot(data=pest_weather, x='Outbreak', y='Rainfall_mm', ax=axes[2], palette='Oranges')
axes[2].set_title('Rainfall (mm) vs Outbreak')
plt.tight_layout()
plt.show()
```

```python
# Pest species distribution across crops
plt.figure(figsize=(12, 6))
sns.countplot(data=pest_weather, x='Crop', hue='Pest_Name', palette='tab10')
plt.title('Pest Species Frequency by Crop')
plt.xticks(rotation=45)
plt.show()
```
